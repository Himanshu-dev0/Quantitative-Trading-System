```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/nifty_regimes_5min.csv", index_col=0, parse_dates=True)

# Strategy Implementation [cite: 87, 102]
df['Signal'] = 0
# Long: EMA Cross + Regime +1 [cite: 88-90]
df.loc[(df['EMA_5'] > df['EMA_15']) & (df['Regime'] == 1), 'Signal'] = 1
# Short: EMA Cross + Regime -1 [cite: 95-97]
df.loc[(df['EMA_5'] < df['EMA_15']) & (df['Regime'] == 2), 'Signal'] = -1 

# Backtest Returns [cite: 104]
df['Market_Ret'] = df['Close'].pct_change()
df['Strategy_Ret'] = df['Signal'].shift(1) * df['Market_Ret']

# Metrics [cite: 107]
sharpe = (df['Strategy_Ret'].mean() / df['Strategy_Ret'].std()) * np.sqrt(252*75)
print(f"Sharpe Ratio: {sharpe}")

df.to_csv("../results/baseline_results.csv") # [cite: 149]
```
