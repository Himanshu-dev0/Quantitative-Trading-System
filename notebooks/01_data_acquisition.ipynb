```python
import yfinance as yf
import pandas as pd
import os

os.makedirs('../data', exist_ok=True)

print("Fetching NIFTY Spot data (last 60 days, 5m)...")

nifty_spot = yf.download(
    "^NSEI",
    period="60d",        # MAX allowed for 5m
    interval="5m",
    progress=False
)

if nifty_spot.empty:
    raise ValueError("No data returned. Yahoo limitation hit.")

nifty_spot.to_csv("../data/nifty_spot_5min.csv")
print("Saved: nifty_spot_5min.csv")
```
